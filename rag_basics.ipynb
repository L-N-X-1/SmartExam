```python
%pip install PyPDF2
%pip install --upgrade langchain
%pip install langchain-text-splitters
%pip install sentence-transformers
%pip install faiss-cpu
%pip install langchain-community




import os
from PyPDF2 import PdfReader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from sentence_transformers import SentenceTransformer
from langchain_community.vectorstores import FAISS
import numpy as np

# Étape 1: Extraction texte (open-source, gère tables/images)
pdf_path = "AI_Workflows__1763328271.pdf"  # remplace par ton PDF
reader = PdfReader(pdf_path)
rreader = PdfReader(pdf_path)
raw_text = ""
for page in reader.pages:
    page_text = page.extract_text()
    if page_text:  # Skip pages vides
        raw_text += page_text + "\n\n"

# Étape 2: Chunking (best practice 2025: recursive + semantic)
splitter = RecursiveCharacterTextSplitter(chunk_size=800, chunk_overlap=150)
chunks = splitter.split_text(raw_text)
print(f"Nombre de chunks: {len(chunks)}")

# Étape 3: Embeddings open-source (BGE-large: top reco 2025)
model = SentenceTransformer("BAAI/bge-large-en-v1.5")  # ou "intfloat/e5-mistral-7b-instruct"
embeddings = model.encode(chunks, batch_size=32, show_progress_bar=True)

# Étape 4: Vector Store (FAISS open-source)
vectorstore = FAISS.from_embeddings(list(zip(chunks, embeddings)), embedding_function=lambda x: model.encode(x))
vectorstore.save_local("data/vector_store/")
print("Vector store saved!")
```
